```python
def clasificacion_empresa(x):
    if x > 100000000:
        return 'Operación > 100 mm'
    elif x > 50000000:
        return 'Operación > 50 mm'
    elif x > 30000000:
        return 'Operación > 30 mm'
    elif x > 20000000:
        return 'Operación > 20 mm'
    elif x > 10000000:
        return 'Operación > 10 mm'
    elif x > 5000000:
        return 'Operación > 5 mm'
    elif x > 2000000:
        return 'Operación > 2 mm'
    elif x > 1000000:
        return 'Operación > 1 mm'
    else:
        return 'Montos menores'
```

```python
Exportaciones['Calificacion tamaño (Exportacion)'] =  Exportaciones['Total FOB'].apply(clasificacion_empresa)
Importaciones['Calificacion tamaño (Importacion)'] = Importaciones['Monto (USD)'].apply(clasificacion_empresa)
```

```python
Exportaciones['Tipo de operación'] = 'Exportación'
Importaciones['Tipo de operación'] = 'Importación'
```

```python
def operaciones_unidos (x):
    if x['Sede de Operaciones'] != 'S/C':
        return x['Sede de Operaciones']
    else:
        return x['Operaciones']
```

```python
def rubros_unidos (x):
    if x['Rubro_x'] != 'S/C':
        return x['Rubro_x']
    else:
        return x['Rubro_y']
```

```python
Empresa['Rubro'] = Empresa.apply(rubros_unidos,  axis = 1)
Empresa['Operaciones Nacionales'] = Empresa.apply(operaciones_unidos,  axis = 1)
```

## San Cristobal

```python
# Lista de partidas (meses)
meses = [
    'en-feb', 'jul-sept', 'mar-jul', 'oct - en26'
]

# Ruta base
base_path = r'D:\Usuarios\RSanabria\OneDrive - HANSA\Escritorio\Business Analytics\Exportaciones importaciones\San Cristobal'

# Lista para almacenar dataframes
dfs = []

for mes in meses:
    file_path = os.path.join(base_path, f"{mes}.xlsx")  # agregar extensión
    
    if os.path.exists(file_path):
        df_temp = pd.read_excel(file_path)
        df_temp["partida"] = mes  # columna identificadora
        dfs.append(df_temp)
        print(f"Archivo cargado: {mes} ✅")
    else:
        print(f"No se encontró el archivo: {mes}")

# Unir todo
if dfs:
    df_sc = pd.concat(dfs, ignore_index=True)
    print("Carga completa ✅")
else:
    print("No se cargó ningún archivo ❌")
```
